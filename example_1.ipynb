**Example 1: $f\widehat{CFD}(S^1\times D^2; C_{1,1};w,\underline{p})$ for the core of the solid torus**

In this example, we compute the filtered bordered invariant associated to the **core of the solid torus**. We start with the filtered type $D$ structure for the solid torus and perform the following operations.

1. Start with the filtered solid-torus type $D$ structure `fCFDsolidtorus_1`, representing the core of $S^1\times D^2$ with the chosen parametrization.

2. Take the associated graded object by setting every positive power of $W_i$ equal to zero. We call the resulting type $D$ structure `gCFDsolidtorus_1`.

3. Convert both the filtered and graded type $D$ structures to type $A$ structures by tensoring with the identity bimodule:
   
   $CFAA(\mathrm{id})\boxtimes CFD.$
   
   This gives `fCFAsolidtorus_1` and `gCFAsolidtorus_1`. After taking the tensor product, we cancel all removable coefficient-$1$ differentials. We keep the surviving generators in their original tensor-product notation; no generator renaming is performed.

4. Tensor the graded type $A$ structure `gCFAsolidtorus_1` with the **unreduced tube-cutting piece**. We first keep the full unreduced product and display all of its generators and type $D$ differential. We then cancel the removable coefficient-$1$ differentials and display the reduced graded product.

5. Repeat the same calculation for the filtered type $A$ structure `fCFAsolidtorus_1`. We first compute and display the **unreduced filtered product** with the tube-cutting piece, and then reduce it by cancelling removable coefficient-$1$ differentials and display the reduced filtered type $D$ structure.

No Dehn twist is used in this example.

Idempotent convention: code idempotent `0` is $\iota_0=\bullet$ (idempotent 1), while code idempotent `1` is $\iota_1=\circ$ (idempotent 2).


**Step 1: Import the commands**


```python
from copy import deepcopy

from IPython.display import display, Markdown, Math

from filtered_multimodules import (
    Multimodule,
    CFAA_id,
    fCFDsolidtorus_1,
    CFDA_twist2,
    tube_cutting_piece,
)

from tensormor import (
    tensor,
    unfiltered_reduced,
    gradedW0,
)

from display_latex import display_type_D_differential
```

**Step 2: Construct the graded solid-torus type $D$ structure**

Start with `fCFDsolidtorus_1()` and set every $W_i=0$ to obtain `gCFDsolidtorus_1`.

No Dehn twist is applied.


```python
gCFDsolidtorus_1 = gradedW0(
    fCFDsolidtorus_1(),
    simplify=True,
)
```

**Step 3: Convert both type $D$ structures to type $A$**

For each solid-torus module we take $CFAA(\mathrm{id})\boxtimes CFD.$

Boundary index `1` of `CFAA_id()` is glued to the only type $D$ boundary. We then cancel removable coefficient-$1$ differentials, exactly as in `example_3`.

The results are named `gCFAsolidtorus_1` and `fCFAsolidtorus_1`. Their surviving generators are kept with the names produced by the tensor product, such as $(x,p)$, $(w_1,p)$, and $(w_2,p)$; no additional relabeling is applied.


```python
gCFAsolidtorus_1 = tensor(
    CFAA_id(),
    1,
    gCFDsolidtorus_1,
    0,
    simplify=False,
)
gCFAsolidtorus_1.remove_differentials()

fCFAsolidtorus_1 = tensor(
    CFAA_id(),
    1,
    fCFDsolidtorus_1(),
    0,
    simplify=False,
)
fCFAsolidtorus_1.remove_differentials()

assert gCFAsolidtorus_1.boundary_types == ["A"]
assert fCFAsolidtorus_1.boundary_types == ["A"]
```

**Step 4: Graded product — unreduced**

Take

\begin{equation*}
gCFA_{\mathrm{solidtorus},1}\boxtimes CFDD_{\mathrm{tube}}.
\end{equation*}

We glue the only type $A$ boundary of `gCFAsolidtorus_1` to boundary index `0` of the **unreduced** `tube_cutting_piece()`. The other tube boundary remains type $D$.

No differential cancellation is performed in this step. The complete generator set and differential are displayed below using the original generator names.


```python
gCFDsolidtorus_1_tube = tensor(
    gCFAsolidtorus_1,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

assert gCFDsolidtorus_1_tube.boundary_types == ["D"]

generators_latex, operations_latex = display_type_D_differential(
    gCFDsolidtorus_1_tube,
    show=False,
)

display(Markdown("**Unreduced graded type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 5: Graded product — reduced**

Make a copy of the unreduced graded product and cancel all removable coefficient-$1$ differentials. The unreduced object is left unchanged, and the surviving generators retain their original names.


```python
gCFDsolidtorus_1_tube_reduced = deepcopy(
    gCFDsolidtorus_1_tube
)
gCFDsolidtorus_1_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    gCFDsolidtorus_1_tube_reduced,
    show=False,
)

display(Markdown("**Reduced graded type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 6: Filtered product — unreduced**

Now take $fCFA_{\mathrm{solidtorus},1}\boxtimes CFDD_{\mathrm{tube}}.$

Again we use the **unreduced** tube-cutting piece and set `simplify=False`, so this cell displays the complete unreduced filtered product before cancellation. The tensor-product generator names are used directly, with no renaming step.


```python
fCFDsolidtorus_1_tube = tensor(
    fCFAsolidtorus_1,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

assert fCFDsolidtorus_1_tube.boundary_types == ["D"]

generators_latex, operations_latex = display_type_D_differential(
    fCFDsolidtorus_1_tube,
    show=False,
)

display(Markdown("**Unreduced filtered type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 7: Filtered product — reduced**

Make a copy of the unreduced filtered product and cancel all removable coefficient-$1$ differentials. Non-unit $W_i$-weighted arrows are retained, and the surviving generators keep their original tensor-product names.


```python
fCFDsolidtorus_1_tube_reduced = deepcopy(
    fCFDsolidtorus_1_tube
)
fCFDsolidtorus_1_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    fCFDsolidtorus_1_tube_reduced,
    show=False,
)

display(Markdown("**Reduced filtered type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```
