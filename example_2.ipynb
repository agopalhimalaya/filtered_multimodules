**Example 2: $f\widehat{CFD}(S^1\times D^2; C_{2,1};w,\underline{p})$ for the $(2,1)$ cable in the solid torus**

In this example, we start with the filtered type $D$ structure for the $(2,1)$ cable in the solid torus and perform the following operations.

1. Define the filtered type $D$ structure `fCFDsolidtorus_21` with generators $a,b,c$ and

\begin{equation*}
\delta(a)=\rho_3\otimes c+W_1\rho_{123}\otimes b+W_1^2\rho_{12}\otimes a,
\end{equation*}

\begin{equation*}
\delta(b)=W_1\otimes c,\qquad \delta(c)=0.
\end{equation*}

2. Display `fCFDsolidtorus_21` using `display_type_D_differential` and check the type $D$ boundary-square condition $d^2=0$.

3. Take the associated graded object by setting every positive power of $W_i$ equal to zero. We call the resulting type $D$ structure `gCFDsolidtorus_21`.

4. Convert both the filtered and graded type $D$ structures to type $A$ structures by tensoring with $CFAA(\mathrm{id})$. We reduce removable coefficient-$1$ differentials and save the results as `fCFAsolidtorus_21` and `gCFAsolidtorus_21`. The surviving generators retain the original tensor-product names produced by the computation; no additional generator relabeling is performed.

5. Tensor the graded type $A$ structure `gCFAsolidtorus_21` with the **unreduced tube-cutting piece**. First display the complete unreduced product, then make a copy, reduce it by cancelling removable coefficient-$1$ differentials, and display the reduced graded product.

6. Repeat the same calculation for the filtered type $A$ structure `fCFAsolidtorus_21`: first compute and display the **unreduced filtered product**, then reduce and display it again.

No Dehn twist is used in this example.

Idempotent convention: code idempotent `0` is $\iota_0=\bullet$ (idempotent 1), while code idempotent `1` is $\iota_1=\circ$ (idempotent 2).


**Step 1: Import the commands**


```python
from copy import deepcopy

from IPython.display import display, Markdown, Math

from filtered_multimodules import (
    Multimodule,
    CFAA_id,
    tube_cutting_piece,
)

from tensormor import (
    tensor,
    unfiltered_reduced,
    gradedW0,
)

from display_latex import display_type_D_differential
```

**Step 2: Define the filtered type $D$ structure for the $(2,1)$ cable in the solid torus**

The idempotents are forced by the Reeb-chord labels:
$a\in\iota_0=\bullet$ and $b,c\in\iota_1=\circ$.

The filtered operations are
\begin{equation*}
a\xrightarrow{\rho_3}c,\qquad
b\xrightarrow{W_1}c,
\end{equation*}

\begin{equation*}
a\xrightarrow{W_1\rho_{123}}b,\qquad
a\xrightarrow{W_1^2\rho_{12}}a.
\end{equation*}

```python
fCFDsolidtorus_21 = Multimodule(
    name="fCFDsolidtorus_21",
    boundary_types=["D"],

    generators=[
        "a",
        "b",
        "c",
    ],

    idempotence={
        "a": [0],
        "b": [1],
        "c": [1],
    },

    operations=[
        ["a", "",      "3",   "c"],
        ["b", "W_1",   "",    "c"],
        ["a", "W_1",   "123", "b"],
        ["a", "W_1^2", "12",  "a"],
    ],
)
```

**Display the filtered type $D$ structure for the $(2,1)$ cable**

```python
generators_latex, operations_latex = display_type_D_differential(
    fCFDsolidtorus_21,
    show=False,
)

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 3: Check the type $D$ boundary-square condition $d^2=0$**

The two nonzero length-two contributions from $a$ to $c$ have coefficient $W_1^2\rho_{123}$ and cancel over $\mathbb{F}_2$.

```python
boundary_square = fCFDsolidtorus_21.d_squared(
    print_result=False
)

if boundary_square == []:
    print("delta^2 = 0")
else:
    print("delta^2 is not zero:")
    for op in boundary_square:
        print(op)
```

**Step 4: Take the associated graded type $D$ structure by setting $W_i=0$**

Every operation carrying a positive power of $W_i$ disappears. The remaining operation is $a\xrightarrow{\rho_3}c$.

We save the result as `gCFDsolidtorus_21`.

```python
gCFDsolidtorus_21 = gradedW0(
    fCFDsolidtorus_21,
    simplify=True,
)

generators_latex, operations_latex = display_type_D_differential(
    gCFDsolidtorus_21,
    show=False,
)

display(Markdown("**Associated graded type $D$ structure, $W_i=0$**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 5: Convert the filtered and graded type $D$ structures to type $A$**

For each module we take
\begin{equation*}
CFAA(\mathrm{id})\boxtimes CFD.
\end{equation*}

Boundary index `1` of `CFAA_id()` is glued to the only type $D$ boundary. The other boundary of $CFAA(\mathrm{id})$ remains type $A$.

After tensoring, we cancel all removable coefficient-$1$ differentials. The surviving generators keep their original tensor-product names; no separate relabeling step is used.


```python
fCFAsolidtorus_21 = tensor(
    CFAA_id(),
    1,
    fCFDsolidtorus_21,
    0,
    simplify=False,
)
fCFAsolidtorus_21.remove_differentials()


gCFAsolidtorus_21 = tensor(
    CFAA_id(),
    1,
    gCFDsolidtorus_21,
    0,
    simplify=False,
)
gCFAsolidtorus_21.remove_differentials()


assert fCFAsolidtorus_21.boundary_types == ["A"]
assert gCFAsolidtorus_21.boundary_types == ["A"]
```

**Step 6: Graded product with the tube-cutting piece — unreduced**

Take
\begin{equation*}
gCFA_{\mathrm{solidtorus},21}\boxtimes CFDD_{\mathrm{tube}}.
\end{equation*}

We glue the only type $A$ boundary of `gCFAsolidtorus_21` to boundary index `0` of the **unreduced** `tube_cutting_piece()`.

No differential cancellation is performed in this step. The complete generator set and differential are displayed below using the original tensor-product generator names.


```python
gCFDsolidtorus_21_tube = tensor(
    gCFAsolidtorus_21,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

assert gCFDsolidtorus_21_tube.boundary_types == ["D"]

generators_latex, operations_latex = display_type_D_differential(
    gCFDsolidtorus_21_tube,
    show=False,
)

display(Markdown("**Unreduced graded type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 7: Reduce the graded product**

Make a copy of the unreduced graded product and cancel all removable coefficient-$1$ differentials. The unreduced object is preserved, and the surviving generators retain their original names.


```python
gCFDsolidtorus_21_tube_reduced = deepcopy(
    gCFDsolidtorus_21_tube
)

gCFDsolidtorus_21_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    gCFDsolidtorus_21_tube_reduced,
    show=False,
)

display(Markdown("**Reduced graded type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 8: Filtered product with the tube-cutting piece — unreduced**

Now take
\begin{equation*}
fCFA_{\mathrm{solidtorus},21}\boxtimes CFDD_{\mathrm{tube}}.
\end{equation*}

Again, the unreduced tube-cutting piece is used and `simplify=False`, so the full filtered product is displayed before cancellation. The original tensor-product generator names are used directly.


```python
fCFDsolidtorus_21_tube = tensor(
    fCFAsolidtorus_21,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

assert fCFDsolidtorus_21_tube.boundary_types == ["D"]

generators_latex, operations_latex = display_type_D_differential(
    fCFDsolidtorus_21_tube,
    show=False,
)

display(Markdown("**Unreduced filtered type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 9: Reduce the filtered product**

Make a copy of the unreduced filtered product and cancel all removable coefficient-$1$ differentials. The $W_i$-weighted arrows are retained unless they participate in the induced zig-zag terms, and the surviving generators keep their original names.


```python
fCFDsolidtorus_21_tube_reduced = deepcopy(
    fCFDsolidtorus_21_tube
)

fCFDsolidtorus_21_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    fCFDsolidtorus_21_tube_reduced,
    show=False,
)

display(Markdown("**Reduced filtered type $D$ product with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```
