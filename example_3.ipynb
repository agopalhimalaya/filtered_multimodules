**Example 3: $f\widehat{CFD}(S^3 \setminus \nu(T_L);\mu;w,\underline{p})$**

In this example, we start with the type $DD$ bimodule for the left-handed trefoil union its meridian given in Jaepil Lee's *Bordered Floer homology and a meridional class of knot* and perform the following operations.

1. Define the type $DD$ bimodule `lhtrefoil_meridian_DD`, corresponding to $\widehat{CFDD}(S^3\setminus\nu(T_L\cup\mu))$ with $n(T_L)=-1$. The first boundary is the $\rho$ boundary and the second boundary is the $\sigma$ boundary.

2. Display `lhtrefoil_meridian_DD` using `display_type_D_differential` and check the type $DD$ boundary-square condition $d^2=0$.

3. Start with the filtered solid-torus type $D$ structure $fCFD_{\mathrm{solidtorus},1}$ and convert it to the filtered type $A$ structure $fCFA_{\mathrm{solidtorus},1}$ by taking $CFAA(\mathrm{id})\boxtimes fCFD_{\mathrm{solidtorus},1}$. We then cancel all removable coefficient-$1$ differentials.

4. Cap off the second, $\sigma$, boundary of `lhtrefoil_meridian_DD` with $fCFA_{\mathrm{solidtorus},1}$. We compute $fCFA_{\mathrm{solidtorus},1}\boxtimes_{\sigma} lhtrefoil\_meridian\_DD$. The $\rho$ boundary remains, so the result is a filtered type $D$ structure `trefoil_meridian_D`. We reduce this structure by cancelling all removable coefficient-$1$ differentials.

5. Change the framing from $n(T_L)=-1$ to $n(T_L)=0$ by taking the box tensor product of `trefoil_meridian_D` with the positive Dehn twist about $\alpha_2$, $CFDA(\tau_{\alpha_2}^{+})$. We glue the type $A$ boundary of the Dehn-twist bimodule to the type $D$ boundary of `trefoil_meridian_D`. The result is `trefoil_meridian_twisted_D`, which we reduce by cancelling all removable coefficient-$1$ differentials.

6. Take the associated graded of `trefoil_meridian_twisted_D` by setting $W_i=0$ using `gradedW0`. We use `simplify=True`, so the resulting graded type $D$ structure `trefoil_meridian_twisted_D_graded` is also reduced.

7. Convert both `trefoil_meridian_twisted_D` and `trefoil_meridian_twisted_D_graded` into type $A$ structures by tensoring with $CFAA(\mathrm{id})$. We reduce both results and save them as `fCFA_lhtrefoil_meridian` and `gCFA_lhtrefoil_meridian`. The surviving generators retain the original tensor-product names produced by the computation; no additional generator renaming is performed.

8. Take the unreduced box tensor product of `gCFA_lhtrefoil_meridian` with the tube-cutting piece $CFDD_{\mathrm{tube}}$. We glue its only type $A$ boundary to the first type $D$ boundary of the tube-cutting piece. The second type $D$ boundary remains. The unreduced result is saved as `gCFD_lhtrefoil_meridian_tube`.

9. Make a copy of `gCFD_lhtrefoil_meridian_tube` and cancel all removable coefficient-$1$ differentials. The reduced graded result is saved as `gCFD_lhtrefoil_meridian_tube_reduced`.

10. Take the unreduced box tensor product of `fCFA_lhtrefoil_meridian` with the tube-cutting piece $CFDD_{\mathrm{tube}}$. Again, we glue its only type $A$ boundary to the first type $D$ boundary of the tube-cutting piece. The second type $D$ boundary remains. The unreduced filtered result is saved as `fCFD_lhtrefoil_meridian_tube`.

11. Make a copy of `fCFD_lhtrefoil_meridian_tube` and cancel all removable coefficient-$1$ differentials. The reduced filtered result is saved as `fCFD_lhtrefoil_meridian_tube_reduced`.

Throughout the calculation, $0=\iota_0=\bullet$ and $1=\iota_1=\circ$. For the initial type $DD$ bimodule, the first boundary uses the $\rho$ algebra and the second boundary uses the $\sigma$ algebra.


**Step 1: Import the commands**

```python
from copy import deepcopy

from IPython.display import display, Markdown, Math

from filtered_multimodules import (
    Multimodule,
    CFAA_id,
    fCFDsolidtorus_1,
    CFDA_twist2,
    tube_cutting_piece,
)

from tensormor import (
    tensor,
    unfiltered_reduced,
    gradedW0,
)

from display_latex import display_type_D_differential
```

**Step 2: Defining the $\widehat{CFDD}(S^3 \setminus v(T_L \cup \mu))$ with $n(T_L)=-1$. From Jaepil Lee's paper!**

```python
lhtrefoil_meridian_DD = Multimodule(
    name="lhtrefoil_meridian_DD",
    boundary_types=["D", "D"],

    generators=[
        "a'",
        "b'",
        "c'",
        "d'",
        "e'",
        "f'",
        "g'",
        "h'",
        "i'",
    ],

    idempotence={
        "a'": [0, 0],
        "b'": [1, 1],
        "c'": [1, 0],
        "d'": [1, 1],
        "e'": [0, 0],
        "f'": [1, 0],
        "g'": [0, 0],
        "h'": [1, 1],
        "i'": [1, 0],
    },

    operations=[
        ["a'", "", "1",   "3",   "b'"],
        ["a'", "", "123", "123", "b'"],

        ["a'", "", "3", "12", "i'"],
        ["i'", "", "2", "",   "a'"],

        ["b'", "", "",   "2", "c'"],
        ["c'", "", "23", "1", "b'"],

        ["c'", "", "",   "1", "d'"],
        ["d'", "", "23", "2", "c'"],

        ["e'", "", "1",   "3",   "d'"],
        ["e'", "", "123", "123", "d'"],

        ["e'", "", "3", "12", "f'"],
        ["f'", "", "2", "",   "e'"],

        ["f'", "", "2", "12", "g'"],
        ["g'", "", "3", "",   "f'"],

        ["g'", "", "1",   "3",   "h'"],
        ["g'", "", "123", "123", "h'"],

        ["h'", "", "23", "2", "i'"],
        ["i'", "", "",   "1", "h'"],
    ],
)
```

**Display the $\widehat{CFDD}(S^3 \setminus v(T_L \cup \mu))$**

```python
generators_latex, operations_latex = display_type_D_differential(
    lhtrefoil_meridian_DD,
    show=False,
)

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 3: Check the $d^2=0$** 

```python
boundary_square = lhtrefoil_meridian_DD.d_squared(print_result=False)

if boundary_square == []:
    print("delta^2 = 0")
else:
    print("delta^2 is not zero:")
    for op in boundary_square:
        print(op)
```

**Step 4: We want to put back the core of the meridian, so we need $fCFA_{\mathrm{solidtorus},1}$. We obtain it by taking the box tensor product $CFAA(\mathbb{I})\boxtimes fCFD_{\mathrm{solidtorus},1}$.**

```python
fCFAsolidtorus_1 = tensor(
    CFAA_id(),
    1,
    fCFDsolidtorus_1(),
    0,
    simplify=False,
)

fCFAsolidtorus_1.remove_differentials()
```

**Step 5: Cap off the second boundary of the left-handed trefoil union meridian $CFDD$ with $fCFA_{\mathrm{solidtorus},1}$.**

We glue the only type $A$ boundary of $fCFA_{\mathrm{solidtorus},1}$ to the second, $\sigma$, boundary of $lhtrefoil\_meridian\_DD$.

The box tensor product is $fCFA_{\mathrm{solidtorus},1}\boxtimes_{\sigma} lhtrefoil\_meridian\_DD$.

After gluing the $\sigma$ boundary, the $\rho$ boundary remains, so the result is a filtered type $D$ structure for the left-handed trefoil complement with the meridian embedded.

```python
trefoil_meridian_D = tensor(
    fCFAsolidtorus_1,
    0,
    lhtrefoil_meridian_DD,
    1,
    simplify=True,
)

trefoil_meridian_D.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    trefoil_meridian_D,
    show=False,
)

display(Markdown("**Reduced filtered type $D$ structure**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 6: Change the framing by applying a positive Dehn twist about $\alpha_2$ to make $n(T_L)=0$.**

We take the box tensor product of the positive Dehn twist $CFDA(\tau_{\alpha_2}^{+})$ with the reduced filtered type $D$ structure `trefoil_meridian_D`.

We glue the type $A$ boundary of $CFDA(\tau_{\alpha_2}^{+})$ to the $\rho$ boundary of `trefoil_meridian_D`. The type $D$ boundary of the Dehn twist remains, so the result is again a filtered type $D$ structure.

We then cancel all removable coefficient-$1$ differentials to obtain the reduced filtered type $D$ structure with the new framing.

```python
trefoil_meridian_twisted_D = tensor(
    CFDA_twist2(),
    1,
    trefoil_meridian_D,
    0,
    simplify=True,
)

trefoil_meridian_twisted_D.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    trefoil_meridian_twisted_D,
    show=False,
)

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 7: Take the associated graded of the reduced filtered type $D$ structure after the positive Dehn twist by setting $W_i=0$, and then reduce it.**

We apply `gradedW0` to `trefoil_meridian_twisted_D`.

Setting $W_i=0$ removes every operation whose coefficient contains a positive power of any $W_i$.

Using `simplify=True` then cancels all removable coefficient-$1$ differentials.

The result is the reduced associated graded type $D$ structure after the framing change.

```python
trefoil_meridian_twisted_D_graded = gradedW0(
    trefoil_meridian_twisted_D,
    simplify=True,
)

generators_latex, operations_latex = display_type_D_differential(
    trefoil_meridian_twisted_D_graded,
    show=False,
)

display(Markdown("**Reduced associated graded type $D$ structure after the positive Dehn twist, $W_i=0$**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 8: Convert both the reduced filtered and reduced graded type $D$ structures into type $A$ structures using $CFAA(\mathrm{id})$.**

For the filtered structure, we take $CFAA(\mathrm{id})\boxtimes trefoil\_meridian\_twisted\_D$ and save the reduced result as `fCFA_lhtrefoil_meridian`.

For the graded structure, we take $CFAA(\mathrm{id})\boxtimes trefoil\_meridian\_twisted\_D\_graded$ and save the reduced result as `gCFA_lhtrefoil_meridian`.

In each case, the second type $A$ boundary of $CFAA(\mathrm{id})$ is glued to the only type $D$ boundary of the corresponding module, leaving one type $A$ boundary.

After reduction, the surviving generators keep the original tensor-product names produced by the computation. No separate relabeling or renaming step is used.


```python
fCFA_lhtrefoil_meridian = tensor(
    CFAA_id(),
    1,
    trefoil_meridian_twisted_D,
    0,
    simplify=False,
)

fCFA_lhtrefoil_meridian.remove_differentials()


gCFA_lhtrefoil_meridian = tensor(
    CFAA_id(),
    1,
    trefoil_meridian_twisted_D_graded,
    0,
    simplify=False,
)

gCFA_lhtrefoil_meridian.remove_differentials()
```

**Step 9: Take the unreduced box tensor product of the graded type $A$ structure with the tube-cutting piece.**

We take the product $gCFA_{\mathrm{lhtrefoil\_meridian}}\boxtimes CFDD_{\mathrm{tube}}$.

We glue the only type $A$ boundary of `gCFA_lhtrefoil_meridian` to the first type $D$ boundary of the tube-cutting piece.

The second type $D$ boundary of the tube-cutting piece remains, so the result is a type $D$ structure.

At this stage, we do not cancel any differentials. The unreduced tensor product is saved as `gCFD_lhtrefoil_meridian_tube`, and the original tensor-product generator names are used directly.


```python
gCFD_lhtrefoil_meridian_tube = tensor(
    gCFA_lhtrefoil_meridian,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

generators_latex, operations_latex = display_type_D_differential(
    gCFD_lhtrefoil_meridian_tube,
    show=False,
)

display(Markdown("**Graded type $D$ structure after tensoring with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 10: Reduce the graded type $D$ structure obtained from the tube-cutting tensor product.**

We make a copy of `gCFD_lhtrefoil_meridian_tube` so that the unreduced tensor product is preserved.

We then cancel all removable coefficient-$1$ differentials in the copy.

The resulting module is saved as `gCFD_lhtrefoil_meridian_tube_reduced`. The surviving generators retain their original names.


```python
gCFD_lhtrefoil_meridian_tube_reduced = deepcopy(
    gCFD_lhtrefoil_meridian_tube
)

gCFD_lhtrefoil_meridian_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    gCFD_lhtrefoil_meridian_tube_reduced,
    show=False,
)

display(Markdown("**Reduced graded type $D$ structure after tensoring with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 11: Take the unreduced box tensor product of the filtered type $A$ structure with the tube-cutting piece.**

We take the product $fCFA_{\mathrm{lhtrefoil\_meridian}}\boxtimes CFDD_{\mathrm{tube}}$.

We glue the only type $A$ boundary of `fCFA_lhtrefoil_meridian` to the first type $D$ boundary of the tube-cutting piece.

The second type $D$ boundary of the tube-cutting piece remains, so the result is a filtered type $D$ structure.

At this stage, we do not cancel any differentials. The unreduced filtered tensor product is saved as `fCFD_lhtrefoil_meridian_tube`, using the original tensor-product generator names.


```python
fCFD_lhtrefoil_meridian_tube = tensor(
    fCFA_lhtrefoil_meridian,
    0,
    tube_cutting_piece(),
    0,
    simplify=False,
)

generators_latex, operations_latex = display_type_D_differential(
    fCFD_lhtrefoil_meridian_tube,
    show=False,
)

display(Markdown("**Filtered type $D$ structure after tensoring with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```

**Step 12: Reduce the filtered type $D$ structure obtained from the tube-cutting tensor product.**

We make a copy of `fCFD_lhtrefoil_meridian_tube` so that the unreduced filtered tensor product is preserved.

We then cancel all removable coefficient-$1$ differentials in the copy.

The resulting reduced filtered type $D$ structure is saved as `fCFD_lhtrefoil_meridian_tube_reduced`. The surviving generators retain their original names.


```python
fCFD_lhtrefoil_meridian_tube_reduced = deepcopy(
    fCFD_lhtrefoil_meridian_tube
)

fCFD_lhtrefoil_meridian_tube_reduced.remove_differentials()

generators_latex, operations_latex = display_type_D_differential(
    fCFD_lhtrefoil_meridian_tube_reduced,
    show=False,
)

display(Markdown("**Reduced filtered type $D$ structure after tensoring with the tube-cutting piece**"))

display(Markdown("Generators"))
display(Math(generators_latex))

display(Markdown("Differential"))
display(Math(operations_latex))
```
